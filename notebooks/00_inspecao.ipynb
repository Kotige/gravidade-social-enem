# Inspeção de todas as tabelas de dados

```python
import duckdb
```

## ENEM 2020

```python

arquivo_2020_microdados = "../data/raw/2020/DADOS/MICRODADOS_ENEM_2020.csv"
arquivo_2020_itens_prova = "../data/raw/2020/DADOS/ITENS_PROVA_2020.csv"
```

```python
duckdb.sql(f"""
    SELECT * FROM read_csv('{arquivo_2020_microdados}', delim=';', encoding='latin-1')
    LIMIT 5
""").show()
```

```python
duckdb.sql(f"""
    SELECT * FROM read_csv('{arquivo_2020_itens_prova}', delim=';', encoding='latin-1')
    LIMIT 5
""").show()
```
